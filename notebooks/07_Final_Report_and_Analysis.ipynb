# 20 — Final Anomaly Analysis

# 20.1 — Anomaly Reason Distribution

```python
# Step 20.1 - Anomaly Reason Distribution

reason_summary = (
    final_anomaly_points["anomaly_reason"]
    .value_counts()
    .rename_axis("anomaly_reason")
    .reset_index(name="count"))

print("========== ANOMALY REASON SUMMARY ==========")
print(reason_summary)
```

# 20.2 — Risk Level Distribution

```python
# Step 20.2 - Risk Level Distribution

risk_summary = (
    final_anomaly_points["final_risk_level"]
    .value_counts()
    .rename_axis("risk_level")
    .reset_index(name="count"))

print("========== ANOMALY POINT RISK SUMMARY ==========")
print(risk_summary)
```

# 20.3 — Patrol-wise Anomaly Points

```python
# Step 20.3 - Patrol-wise Anomaly Points

patrol_anomaly_summary = (
    final_anomaly_points["patrol_id"]
    .value_counts()
    .rename_axis("patrol_id")
    .reset_index(name="anomaly_points"))

print("========== PATROL-WISE ANOMALY SUMMARY ==========")
print(patrol_anomaly_summary)
```

# 20.4 — Model Agreement

```python
# Step 20.4 - Rule vs ML Agreement

agreement_summary = (
    final_anomaly_points["detection_category"]
    .value_counts()
    .rename_axis("detection_category")
    .reset_index(name="count"))

agreement_summary["percentage"] = (
    agreement_summary["count"]
    / len(final_anomaly_points)
    * 100)

print("========== RULE vs ML SUMMARY ==========")
print(agreement_summary)
```

# 20.5 — Speed and Distance Analysis

```python
# Step 20.5 - Movement Statistics

print("========== MOVEMENT STATISTICS ==========")

print(
    "Maximum speed:",
    final_anomaly_points["speed_kmh"].max(),
    "km/h")

print(
    "Maximum distance jump:",
    final_anomaly_points["distance_from_previous_m"].max(),
    "meters")

print(
    "Maximum distance from road:",
    final_anomaly_points["distance_to_road_m"].max(),
    "meters")

print(
    "Maximum hybrid score:",
    final_anomaly_points["final_hybrid_score"].max())
```

# 20.6 — Strongest Anomalies

```python
# Step 20.6 - Top Anomalies
top_anomalies = (
    final_anomaly_points[
        [
            "patrol_id",
            "timestamp",
            "speed_kmh",
            "distance_from_previous_m",
            "distance_to_road_m",
            "final_hybrid_score",
            "final_risk_level",
            "anomaly_reason"
        ]
    ]
    .sort_values(
        "final_hybrid_score",
        ascending=False
    )
    .head(10))

print("========== TOP 10 ANOMALIES ==========")
print(top_anomalies)
```

# 20.7 — Final Actionable Event Analysis

```python
# Step 20.7 - Final Actionable Events

actionable_summary = final_event_dataset[
    [
        "patrol_id",
        "event_id",
        "event_risk_level",
        "event_classification",
        "max_speed_kmh",
        "max_distance_jump_m",
        "max_distance_to_road_m",
        "max_hybrid_score",
        "description",
        "alert_status"
    ]
].sort_values(
    "max_hybrid_score",
    ascending=False)

print("========== ACTIONABLE EVENTS ==========")
print(actionable_summary.to_string(index=False))
```
